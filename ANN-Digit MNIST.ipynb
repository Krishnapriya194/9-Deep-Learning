```python
#Encoding Digit MNIST Labels (One hot encoding)
y_train_digit = to_categorical(y_train_digit, num_classes=10)

y_test_digit = to_categorical(y_test_digit, num_classes=10)
```

```python
#Creating base neural network
model = keras.Sequential([
    layers.Dense(256, activation='relu', input_shape=(784,)),
    #layers.Dropout(0.3),
    #layers.BatchNormalization(),
    layers.Dense(64, activation='relu'),
    #layers.Dropout(0.3),
    #layers.BatchNormalization(),
    layers.Dense(64, activation='relu'),
    #layers.Dropout(0.3),
    #layers.BatchNormalization(),
    layers.Dense(10,activation='sigmoid'),
])
```

```python
#Creating base neural network
model = keras.Sequential([
    Input(shape=(784,)),
    layers.Dense(256, activation='relu'),
    #layers.Dropout(0.3),
    #layers.BatchNormalization(),
    layers.Dense(64, activation='relu'),
    #layers.Dropout(0.3),
    #layers.BatchNormalization(),
    layers.Dense(64, activation='relu'),
    #layers.Dropout(0.3),
    #layers.BatchNormalization(),
    layers.Dense(10,activation='sigmoid'),
])
```

```python
model.summary()
```

https://keras.io/api/losses/probabilistic_losses/#sparsecategoricalcrossentropy-class

"model.compile() configures the neural network before training. 
categorical_crossentropy is used as the loss function because this is a multi-class classification problem with one-hot encoded labels. 
    Adam is chosen as the optimizer to efficiently update the model weights during training, 
and accuracy is used as the evaluation metric to monitor the model's performance after each epoch."


```python
#Compiling the model
model.compile(loss="categorical_crossentropy",
              optimizer="adam",
              metrics = ['accuracy'])
```

```python
history=model.fit(X_train_digit, y_train_digit, batch_size=100, epochs=10,validation_data=(X_test_digit, y_test_digit))
# validation_data tells Keras:
"After each training epoch, evaluate the model on this separate dataset, but don't use it to update the model's weights."means:
✅ Use X_train_fashion and y_train_fashion to train the model.
✅ After each epoch, evaluate the model using X_test_fashion and y_test_fashion.
✅ Report val_loss and val_accuracy.
✅ Do not update the model's weights using the validation data.
✅ Helps detect overfitting and monitor how well the model generalizes to unseen data.

This is one of the key ideas in deep learning: training data teaches the model;
validation data checks how well it has learned without influencing the learning process.
```

```python
test_loss_digit, test_acc_digit = model.evaluate(X_test_digit, y_test_digit)
```

```python
print('Digit MNIST Test accuracy:', round(test_acc_digit,4))
```

```python
#Predicting the labels-DIGIT
y_predict = model.predict(X_test_digit)
#The model does not return the class name directly.
Instead, it returns the probability of each class.
```

Learnings from the confusion matrix?"

"The confusion matrix compares actual labels with predicted labels. The diagonal elements represent correctly classified samples, while the off-diagonal elements represent misclassifications. By examining it, we can identify which classes the model predicts well and which pairs of classes are frequently confused. This helps us understand the model's strengths and weaknesses beyond overall accuracy."

Why use a Heatmap?
draws the confusion matrix visually.
Darker cells (depending on the colormap) indicate larger counts.
annot=True prints the actual numbers in each cell.
fmt="d" displays them as integers.

```python
#Confusion matrix for Digit MNIST
con_mat=confusion_matrix(y_test_digit_eval,y_predicts)
plt.style.use('seaborn-v0_8-deep')
plt.figure(figsize=(10,10))
sns.heatmap(con_mat,annot=True,annot_kws={'size': 15},linewidths=0.5,fmt="d",cmap="gray")
plt.title('True or False predicted digit MNIST\n',fontweight='bold',fontsize=15)
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test_digit_eval,y_predicts))
```

Report tells you how well the model performed for each class, not just the overall accuracy.

"The model achieved an overall accuracy of 89% on the MNIST handwritten digit dataset.
It performs very well on digits 0, 1, 4, 5, 8, and 9, where both precision and recall are high. However, 
the model has relatively lower precision for digit 2, meaning it predicts some other digits as 2, and lower recall for digits 3, 6, and 7, 
indicating that it misses a noticeable number of those digits. The macro and weighted averages are nearly identical 
because the dataset is balanced across all digit classes."
